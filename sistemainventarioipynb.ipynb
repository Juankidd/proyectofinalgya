```python


def registrar_producto():
  nuevo_valor = ""
  nueva_cantidad = 0
  estado = False

  producto = input("Ingrese el nombre del producto: ").strip().lower()
  cantidad = int(input("Ingrese la cantidad del producto: "))

  try:
    with open("inventario.txt","r",encoding="utf-8") as archivo:
      for item in archivo:
        nombre, c = item.split(":")

        if nombre == producto:
          nueva_cantidad = int(c) + cantidad
          nuevo_valor += f"{nombre}:{nueva_cantidad}\n"
          print(nuevo_valor)
          estado = True

        else:
          nuevo_valor += item #Memoria

  except FileNotFoundError:
    print("El archivo no existe, voy a generarlo")

  if not estado:
      nuevo_valor += f"{producto}:{cantidad}\n+" #Agregar un valor nuevo

  else:
      print("Los datos de cantidad han sido actualizados")

#sobrescribir lo anterior con lo actual
  with open("inventario.txt","w",encoding="utf-8") as archivo:
    archivo.write(nuevo_valor)


def buscar_producto():
  estado = False
  producto = input("Ingrese el nombre del producto a buscar: ").strip().lower()

  try:
    with open("inventario.txt","r",encoding="utf-8") as archivo:
      for item in archivo:
        nombre, c = item.split(":")

        if nombre == producto:
          print(f"El producto fue encontrado: {nombre}:{c}")
          estado = True
          break

  except FileNotFoundError:
    print("El invenitario no existe")

  if not estado:
    print("El producto no se encuentra en el inventario")


def ver_inventario():

  try:
    with open("inventario.txt","r",encoding="utf-8") as archivo:
      datos = archivo.read() #Read es para leer todos los valores del archivo
      print("***Tu invenitario***\n")
      print(datos)
      print("********************")

  except FileNotFoundError:
    print("El invenitario no existe")



def menu():
  while True:
    print("---MiniMarket Cali---")
    print("1. Registrar producto")
    print("2. Buscar producto")
    print("3. Ver inventario")
    print("4. Salir")

    opcion =input("Seleccione una opción: \n")

    if opcion == "1":
      registrar_producto()

    elif opcion == "2":
      buscar_producto()

    elif opcion == "3":
      ver_inventario()

    elif opcion == "4":
      print("Gracias por utilizar nuestros servicios en el minimarket Cali")
      break

    else:
      print("La opción es inválida")


#Programa principal
menu()

```
